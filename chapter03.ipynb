### 리스트 메서드

### 튜플 (Tuple)
 - 한번 생성하면 요소를 추가, 수정, 삭제 할 수 없다.

`````python
axis = (38.123, 41.123)

# axis[0] = 36.1234 바뀌지않은 값이라 에러가 뜸

point = 1, 2, 3
print(point, type(point))
`````

`````python
import sys
print(sys.getsizeof([1, 2, 3]))
print(sys.getsizeof((1, 2, 3))) # 튜플 : 테이터 안정성
`````

### 슬라이싱


`````python
numbers = [0, 1, 2, 3, 4,5]

# 1번 부터 4번 미만까지
numbers[1:4]
# 2칸 간격으로 추출
numbers[::2]
# 역순으로 추출
numbers[::-1]
# 앞에서 3개만 추출
numbers[:3]

numbers[3:]
`````

`````python
print ((10, 20, 30, 40)[1:3])
`````

`````python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

print (matrix[1])
print (matrix[1][2])
print (len(matrix), len(matrix[0]))
`````
