## 숫자 맞추기 게임

```python
import random
number = random.randint(1, 10)

print(f"="*40)
print(f"        1 ~ 100 숫자 맞추기 게임!")
print(f"="*40)

try_count = 0


while True:

    user_input = input("숫자를 맞춰보세요!")
    
    if user_input == "q":
        print(f"게임을 종료합니다 정답은 {number} 입니다")
        break

    if not user_input.isdecimal():
        continue

    user_number = int(user_input)
    
    try_count += 1

    if user_number < 0 or user_number > 100:
        print(f"정답은 1 ~ 100 사이에 있습니다")
        continue

    if user_number > number:
        print("정답보다 작은 수입니다")

    elif user_number < number:
        print("정답보다 큰 수입니다")
    
    else :
        print(f"정답입니다 {try_count}회만에 맞추셨습니다")
        break
    
    
    






```
